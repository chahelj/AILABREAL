```python
import time
from collections import deque

N = 3

class PuzzleState:
    def __init__(self, board, x, y, depth):
        self.board = board
        self.x = x
        self.y = y
        self.depth = depth

row = [0, 0, -1, 1]
col = [-1, 1, 0, 0]

def is_goal_state(board):
    goal = [[1, 2, 3], [4, 5, 6], [7, 8, 0]]
    return board == goal

def is_valid(x, y):
    return 0 <= x < N and 0 <= y < N

def print_board(board, indent_level=0):
    indent_str = '  ' * indent_level
    for r_val in board:
        print(f"{indent_str}{' '.join(map(str, r_val))}")
    print(f"{indent_str}--------")

def solve_puzzle_dfs(start, x, y, max_depth):
    stack = []
    visited = set()

    stack.append(PuzzleState(start, x, y, 0))
    visited.add(tuple(map(tuple, start)))

    while stack:
        curr = stack.pop()

        if curr.depth > max_depth:
            continue

        indent_str = '  ' * curr.depth
        print(f"{indent_str}Depth: {curr.depth}")
        print_board(curr.board, curr.depth)

        if is_goal_state(curr.board):
            print(f"{indent_str}Goal state reached at depth {curr.depth}")
            return

        for i in range(4):
            new_x = curr.x + row[i]
            new_y = curr.y + col[i]

            if is_valid(new_x, new_y):
                new_board = [row[:] for row in curr.board]
                new_board[curr.x][curr.y], new_board[new_x][new_y] = new_board[new_x][new_y], new_board[curr.x][curr.y]

                board_tuple = tuple(map(tuple, new_board))
                if board_tuple not in visited:
                    visited.add(board_tuple)
                    stack.append(PuzzleState(new_board, new_x, new_y, curr.depth + 1))

    print(f"No solution found within depth limit {max_depth}")
    print

if __name__ == '__main__':
    print("CHAHEL J 1WN24CS066")
    start = [[1, 2, 3], [0, 4, 6], [7, 5, 8]]

    initial_x, initial_y = -1, -1
    for r_idx in range(N):
        for c_idx in range(N):
            if start[r_idx][c_idx] == 0:
                initial_x, initial_y = r_idx, c_idx
                break
        if initial_x != -1:
            break

    x, y = initial_x, initial_y
    max_depth = 3

    print('Initial State:')
    print_board(start, 0)

    print(f'Attempting to solve with DFS up to depth {max_depth}...')

    start_time = time.time()
    solve_puzzle_dfs(start, x, y, max_depth)
    end_time = time.time()

    print(f"Time taken for DFS: {end_time - start_time:.4f} seconds")
```
