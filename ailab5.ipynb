### 8 Queens Problem using Simulated Annealing

**The Goal:** Place 8 chess queens on an $8 \times 8$ chessboard such that no two queens threaten each other. This means no two queens can share the same row, column, or diagonal.

#### Simulated Annealing formulation:
- **State Representation:** We represent the board as a 1D array of 8 integers: `[r0, r1, ..., r7]`, where the index represents the column and the value represents the row of the queen in that column. This naturally avoids column conflicts.
- **Cost Function (Energy):** The number of conflicting pairs (queens threatening each other). The global minimum (solution) has a cost of $0$.
- **Neighbor Generation:** Choose a random column and change its queen's row to a random new row.
- **Acceptance Probability:** We accept a neighbor state if it's better (lower cost). If it's worse, we accept it with a probability of $e^{-\Delta E / T}$, where $T$ is the current temperature. This temperature gradually decreases ("cools") over iterations.

```python
import random
import math
import numpy as np

def calculate_conflicts(state):
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j]:
                conflicts += 1
            elif abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_neighbor(state):
    neighbor = list(state)
    col = random.randint(0, len(state) - 1)
    current_row = state[col]
    new_row = random.choice([r for r in range(len(state)) if r != current_row])
    neighbor[col] = new_row
    return neighbor

def simulated_annealing(n=8, initial_temp=10.0, cooling_rate=0.995, max_iterations=5000):
    current_state = [random.randint(0, n - 1) for _ in range(n)]
    current_cost = calculate_conflicts(current_state)
    temp = initial_temp
    for i in range(max_iterations):
        if current_cost == 0:
            return current_state, i
        neighbor = get_neighbor(current_state)
        neighbor_cost = calculate_conflicts(neighbor)
        delta_e = neighbor_cost - current_cost
        if delta_e < 0 or random.random() < math.exp(-delta_e / temp):
            current_state = neighbor
            current_cost = neighbor_cost
        temp *= cooling_rate
    return current_state, max_iterations
```

```python
def print_board(state):
    n = len(state)
    for r in range(n):
        row_str = ""
        for c in range(n):
            if state[c] == r:
                row_str += " Q "
            else:
                row_str += " . "
        print(row_str)

solution, iterations = simulated_annealing()
print("chahel J")
print(f"Final State representation: {solution}")
print(f"Remaining conflicts: {calculate_conflicts(solution)}")
print(f"Iterations taken: {iterations}\n")
print("Visual Board Layout:")
print_board(solution)
```
