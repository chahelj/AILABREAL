```python
def print_board(board):
    """Displays the current board state."""
    for row in board:
        print(" | ".join(row))
        print("-" * 9)


def check_winner(board):
    """Checks for a winner. Returns 'X', 'O', 'Tie', or None."""
    # Check rows and columns
    for i in range(3):
        if board[i][0] == board[i][1] == board[i][2] != " ":
            return board[i][0]
        if board[0][i] == board[1][i] == board[2][i] != " ":
            return board[0][i]

    # Check diagonals
    if board[0][0] == board[1][1] == board[2][2] != " ":
        return board[0][0]
    if board[0][2] == board[1][1] == board[2][0] != " ":
        return board[0][2]

    # Check for tie
    if all(cell != " " for row in board for cell in row):
        return "Tie"

    return None


def minimax(board, depth, is_maximizing,SYMBOL):
    """Brute-force explores all possibilities to find the best score."""
    result = check_winner(board)
    if SYMBOL == "X":
        NOTSYMBOL = "O"
    else:
        NOTSYMBOL = "X"
    if result == SYMBOL:  # AI wins
        return 1
    if result == NOTSYMBOL:  # Human wins
        return -1
    if result == "Tie":
        return 0

    if is_maximizing:
        best_score = float("-inf")
        for r in range(3):
            for c in range(3):
                if board[r][c] == " ":
                    board[r][c] = "O"
                    score = minimax(board, depth + 1, False,SYMBOL)
                    board[r][c] = " "  # Undo move
                    best_score = max(score, best_score)
        return best_score
    else:
        best_score = float("inf")
        for r in range(3):
            for c in range(3):
                if board[r][c] == " ":
                    board[r][c] = "X"
                    score = minimax(board, depth + 1, True,SYMBOL)
                    board[r][c] = " "  # Undo move
                    best_score = min(score, best_score)
        return best_score


def find_best_move(board, SYMBOL):
    """Finds the optimal move for the AI."""
    best_score = float("-inf")
    move = (-1, -1)

    for r in range(3):
        for c in range(3):
            if board[r][c] == " ":
                board[r][c] = SYMBOL
                score = minimax(board, 0, False,SYMBOL)
                board[r][c] = " "  # Undo move
                if score > best_score:
                    best_score = score
                    move = (r, c)
    return move


def play_game():
    board = [["O", " ", "X"], ["X", " ", " "], ["X", "O", "O"]]
    print_board(board)

    while True:
        if check_winner(board):
            break

        # AI 1 Turn
        print("AI 1 is making its move...")
        ai_row, ai_col = find_best_move(board, "X")
        if ai_row != -1 and ai_col != -1:
            board[ai_row][ai_col] = "X"
        print_board(board)

        if check_winner(board):
            break

        # AI 2 Turn
        print("AI 2 is making its move...")
        ai_row, ai_col = find_best_move(board, "O")
        if ai_row != -1 and ai_col != -1:
            board[ai_row][ai_col] = "O"
        print_board(board)

        if check_winner(board):
            break

    # Game Over
    winner = check_winner(board)
    if winner == "Tie":
        print("It's a tie!")
    else:
        print(f"{winner} wins!")


if __name__ == "__main__":
    play_game()
```

```python
import time

class VacuumCleaner:
    CLEAN = 'Clean'
    DIRTY = 'Dirty'

    def __init__(self):
        # Initialize user input variables
        self.room_a_status = self.DIRTY
        self.room_b_status = self.DIRTY
        self.current_location = 'A'

    def draw_environment(self):
        print(f"Room A: {self.room_a_status}  Room B: {self.room_b_status}")

    def scan_and_clean(self, room):
        status = getattr(self, f"room_{room.lower()}_status")
        if status == self.DIRTY:
            setattr(self, f"room_{room.lower()}_status", self.CLEAN)
            return True
        else:
            return False

    def move_left(self):
        self.current_location = 'A'

    def move_right(self):
        self.current_location = 'B'

    def run(self):
        print("Initial environment (Before cleaning):")
        self.draw_environment()
        while any(status == self.DIRTY for status in [self.room_a_status, self.room_b_status]):
            while self.scan_and_clean(self.current_location):
                print(f"\nCleaning in Room {self.current_location}...")
                time.sleep(1)
                self.draw_environment()

            if self.current_location == 'A':
                self.move_right()
            else:
                self.move_left()

            print(f"\nMoving to Room {self.current_location}...")
            time.sleep(1)
            self.draw_environment()

        print("\nFinal environment (After cleaning):")
        self.draw_environment()
        print("\nAll rooms are clean. Task completed.")

if __name__ == "__main__":
    vacuum = VacuumCleaner()
    vacuum.run()
```

```python
import copy

def check_winner(board):
    for row in board:
        if row[0] == row[1] == row[2] and row[0] != " ":
            return row[0]

    for col in range(3):
        if board[0][col] == board[1][col] == board[2][col] and board[0][col] != " ":
            return board[0][col]

    if board[0][0] == board[1][1] == board[2][2] and board[0][0] != " ":
        return board[0][0]
    if board[0][2] == board[1][1] == board[2][0] and board[0][2] != " ":
        return board[0][2]

    if all(cell != " " for row in board for cell in row):
        return "Tie"

    return None

def generate_tree(board, turn):
    winner = check_winner(board)
    if winner:
        score_val = 1 if winner == "X" else (-1 if winner == "O" else 0)
        return {"board": board, "score": score_val, "minimax_score": score_val, "children": []}

    children = []
    if turn == "X":
        current_node_minimax_score = float('-inf')
    else:
        current_node_minimax_score = float('inf')

    for r in range(3):
        for c in range(3):
            if board[r][c] == " ":
                new_board = copy.deepcopy(board)
                new_board[r][c] = turn
                next_turn = "O" if turn == "X" else "X"
                child_node = generate_tree(new_board, next_turn)
                children.append(child_node)

                if turn == "X":
                    current_node_minimax_score = max(current_node_minimax_score, child_node["minimax_score"])
                else:
                    current_node_minimax_score = min(current_node_minimax_score, child_node["minimax_score"])

    return {"board": board, "turn": turn, "children": children, "minimax_score": current_node_minimax_score}

def print_tree(node, depth=0):
    indent = "    " * depth

    board_display_lines = []
    for row in node["board"]:
        board_display_lines.append(" | ".join([c if c != " " else "." for c in row]))

    if not node["children"]:
        score_str = f" -> End State (Score: {node['score']}, Minimax Cost: {node['minimax_score']})"
        print(f"{indent}└── {score_str}")
    else:



        print(f"{indent}├── (Next Turn: {node['turn']}, Minimax Cost: {node['minimax_score']})")

    for i, line in enumerate(board_display_lines):
        prefix = f"{indent}    "











    for child in node["children"]:
        print_tree(child, depth + 1)

initial_board = [['O', ' ', 'X'], ['X', ' ', ' '], ['X', 'O', 'O']]

print("Generating game tree... (this takes a few seconds)")
game_tree = generate_tree(initial_board, "X")

print("\nGame Tree Structure:")
print_tree(game_tree)
```
