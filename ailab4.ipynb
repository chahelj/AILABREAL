```python
import random

def get_conflicts(state):
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j]:
                conflicts += 1
            elif abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_best_neighbor(state):
    n = len(state)
    current_conflicts = get_conflicts(state)
    best_state = list(state)
    best_conflicts = current_conflicts
    neighbors = []

    for i in range(n):
        for j in range(i + 1, n):
            neighbor = list(state)
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
            conflicts = get_conflicts(neighbor)

            if conflicts < best_conflicts:
                best_conflicts = conflicts
                neighbors = [neighbor]
            elif conflicts == best_conflicts:
                neighbors.append(neighbor)

    if neighbors:
        return random.choice(neighbors), best_conflicts
    return state, current_conflicts

def hill_climbing_4_queens_path():
    n = 4
    state = [3, 1, 2, 0]
    path = [list(state)]

    while True:
        neighbor, neighbor_conflicts = get_best_neighbor(state)
        if neighbor_conflicts >= get_conflicts(state):
            break
        state = neighbor
        path.append(list(state))

    return path, get_conflicts(state) == 0

def print_board(state):
    n = len(state)
    for row in range(n):
        line = ""
        for col in range(n):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)

path, success = hill_climbing_4_queens_path()
print("Search Path:")
for step, state in enumerate(path):
    print(f"Step {step}: {state} (Conflicts: {get_conflicts(state)})")
print("chahel J")
print(f"\nSuccess: {success}")
print("\nFinal Board State:")
print_board(path[-1])
```
